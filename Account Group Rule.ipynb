```python
row = df2.iloc[10]
account_type      = row["accountMT4Type_display"]
currency_type     = row["currency"]
group_type        = row["group"]

has_currency = currency_type in group_type
has_p  = 'P' in group_type
has_nm = 'NM' in group_type
has_ns = 'NS' in group_type

if "Swap free ECN" in account_type:
    print('Swap free ECN boolean test: ',has_currency and has_p and has_nm)
elif "ECN" in account_type:
    print('ECN boolean test: ',has_currency and has_p)
elif "Swap free STP" in account_type:
    print('Swap free STP boolean test: ',has_currency and not has_p and has_nm)
elif "STP" in account_type:
    print('STP boolean test: ',has_currency and not has_p)
```

```python
import re
def parse_group(group):
    tokens = re.split(r'[_\\]', str(group))  # 同時切 _ 和 \
    return set(filter(None, tokens))        # 去掉空字串

def group_checking(row):
    account_type  = str(row["accountMT4Type_display"])
    currency_type = str(row["currency"])
    group_tokens  = str(row["group"])#parse_group(row["group"])

    has_currency = currency_type in group_tokens
    has_p  = "P"  in group_tokens
    has_nm = "NM" in group_tokens
    has_ns = "NS" in group_tokens
    has_bc = "BC" in group_tokens
    has_r  = "R"  in group_tokens
    has_tv = "TV" in group_tokens
    
    if "Swap free ECN" in account_type:
        return has_currency and has_p and has_nm
    elif "ECN" in account_type:
        return has_currency and has_p

    elif "Swap free STP" in account_type:
        return has_currency and not has_p and has_nm
    elif "STP" in account_type:
        return has_currency and not has_p

    elif "SF Perpetual" in account_type:
        return has_currency and has_bc and has_nm
    elif "Perpetual" in account_type:
        return has_currency and has_bc
        
    elif "SF Premium" in account_type:
        return has_currency and has_r and has_nm
    elif "Premium" in account_type:
        return has_currency and has_r 
        
    elif "Trading View" in account_type:
        return has_currency and has_tv
    
    return False
```

```python
df2["Group Checking"] = df2.apply(group_checking, axis=1)
df2
```

```python
df3 = df2[df2['Group Checking'] == False]
df3['accountMT4Type_display'].unique()
```
